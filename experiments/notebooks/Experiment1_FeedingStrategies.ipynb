```python
import os
import re
import warnings
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

warnings.filterwarnings("ignore")

# --- STEP 1: LOAD AND MERGE BOTH EXPERIMENT RUNS ---
path_feeding = "../results/exp_feeding_strategies.csv"
path_width_depth = "../results/exp_width_depth_scenarios.csv"

# Fallback to local directory if relative path differs
if not os.path.exists(path_feeding):
    path_feeding = "exp_feeding_strategies.csv"
    path_width_depth = "exp_width_depth_scenarios.csv"

print("Loading data files...")
df_feeding = pd.read_csv(path_feeding)
df_width_depth = pd.read_csv(path_width_depth)

# Stack datasets cleanly
df = pd.concat([df_feeding, df_width_depth], ignore_index=True)

# Extract core strategy names by removing width (_mX) and depth (_LX) suffixes
df["strategy_raw"] = df["model"].apply(lambda x: re.sub(r"_m\d+_L\d+$", "", x))


# Map technical identifiers to the corresponding experimental configurations
def assign_formal_scenario(row):
    strat = row["strategy_raw"]
    m_val = row["m"]

    strategy_map = {
        # Experiment 1 Scenarios
        "baseline_fullinput": (
            "A — baseline (full, m=1)" if m_val == 1 else "A — full/full, m>1"
        ),
        "input_subspace_sqrt": "D-√d (input_subspace_sqrt)",
        "input_subspace_dm": "D-d/m (input_subspace_dm)",
        "featpart_fulldata": "B-Φ-P/m (featpart_fulldata)",
        # Experiment 2 Scenarios
        "C_splitdata_fullcols": "C — split/full-cols",
        "B_in_sqrt": "B-input-√d-",
        "B_in_dm": "B-input-d/m",
        "D_sqrt_fullP": "D_sqrt_fullP",
        "D_dm_fullP": "D_dm_fullP",
    }
    return strategy_map.get(strat, f"Unknown ({strat})")


df["exp_name"] = df.apply(assign_formal_scenario, axis=1)


# Helper function to combine mean and standard deviation beautifully
def format_mean_std(series):
    if len(series) == 0:
        return "-"
    mu = series.mean()
    sigma = series.std()
    if pd.isna(sigma):  # If only 1 observation exists
        return f"{mu:.4f} ± 0.0000"
    return f"{mu:.4f} ± {sigma:.4f}"


# --- STEP 2: SUMMARY TABLES WITH STD (ONE PER DATASET) ---
datasets = sorted(df["dataset"].unique())

print("=" * 85)
print("TABLE 1: OVERALL STRATEGY PERFORMANCE BREAKDOWN PER DATASET")
print("=" * 85)

for dataset in datasets:
    print(f"\n>>> DATASET: {dataset.upper()} ")
    print("-" * 50)

    # Filter to isolate current dataset
    dataset_subset = df[df["dataset"] == dataset]

    # Render localized pivot table
    pivot_dataset = dataset_subset.pivot_table(
        index="exp_name", columns="m", values="acc", aggfunc=format_mean_std
    )
    display(pivot_dataset)

print("\n" + "=" * 85)
print("TABLE 2: DETAILED DATASET SCALING WITH DATA & WIDTH (ALL COMBINED)")
print("=" * 85)
pivot_detailed = df.pivot_table(
    index=["dataset", "n_train"],
    columns=["m", "exp_name"],
    values="acc",
    aggfunc=format_mean_std,
)
display(pivot_detailed)

# --- STEP 3: PLOT DATA SCALING FOR EACH DATASET ---
for dataset in datasets:
    dataset_df = df[df["dataset"] == dataset]

    # Dynamically find the widths present for this specific dataset
    m_values = sorted(dataset_df["m"].unique())

    fig, axes = plt.subplots(
        1, len(m_values), figsize=(5.5 * len(m_values), 5), sharey=True
    )
    if len(m_values) == 1:
        axes = [axes]

    fig.suptitle(
        f"Data Scaling Analysis & Strategy Performance: {dataset}",
        fontsize=14,
        fontweight="bold",
        y=1.05,
    )

    for ax, m in zip(axes, m_values):
        sub_df = dataset_df[dataset_df["m"] == m]
        sub_df = sub_df.sort_values(by="n_train")

        sns.lineplot(
            data=sub_df,
            x="n_train",
            y="acc",
            hue="exp_name",
            marker="o",
            ax=ax,
            err_style="band",
            errorbar="sd",  # Using the modern errorbar syntax to prevent deprecation warnings
        )

        ax.set_title(f"Width (m = {m})", fontsize=12, fontweight="semibold")
        ax.set_xlabel("Training Samples (n_train)")
        ax.set_ylabel("Accuracy")
        ax.grid(True, linestyle="--", alpha=0.5)
        ax.set_xscale("log")

        if ax.get_legend() is not None:
            ax.get_legend().remove()

    # Create a uniform global legend on the far right of the subplot string
    handles, labels = axes[0].get_legend_handles_labels()
    fig.legend(
        handles,
        labels,
        loc="center left",
        bbox_to_anchor=(1.01, 0.5),
        title="All Strategy Runs",
    )

    plt.tight_layout()
    plt.show()
```

```python
import os
import re
import warnings
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

warnings.filterwarnings("ignore")

# --- STEP 1: LOAD AND MERGE DATA ---
path_feeding = "../results/exp_feeding_strategies.csv"
path_width_depth = "../results/exp_width_depth_scenarios.csv"

if not os.path.exists(path_feeding):
    path_feeding = "exp_feeding_strategies.csv"
    path_width_depth = "exp_width_depth_scenarios.csv"

df_feeding = pd.read_csv(path_feeding)
df_width_depth = pd.read_csv(path_width_depth)
df = pd.concat([df_feeding, df_width_depth], ignore_index=True)

# Extract core strategy names
df["strategy_raw"] = df["model"].apply(lambda x: re.sub(r"_m\d+_L\d+$", "", x))

# --- STEP 2: ISOLATE AND MAP THE 'D' STRATEGIES FOR ABLATION ---
ablation_map = {
    "input_subspace_sqrt": ("D (√d)", "Old: P/m Budget"),
    "input_subspace_dm":   ("D (d/m)", "Old: P/m Budget"),
    "D_sqrt_fullP":        ("D (√d)", "New: Full P Budget"),
    "D_dm_fullP":          ("D (d/m)", "New: Full P Budget")
}

# Filter only the relevant D strategies
df_ablation = df[df["strategy_raw"].isin(ablation_map.keys())].copy()

# Apply mapping to create two new columns
df_ablation["Subspace_Strategy"] = df_ablation["strategy_raw"].map(lambda x: ablation_map[x][0])
df_ablation["Feature_Budget"] = df_ablation["strategy_raw"].map(lambda x: ablation_map[x][1])

# Filter for m > 1 since m=1 has the exact same budget (1000/1 = 1000) for both
df_ablation = df_ablation[df_ablation["m"] > 1]

# Helper formatter
def format_mean_std(series):
    if len(series) == 0: return "-"
    mu = series.mean()
    sigma = series.std()
    if pd.isna(sigma): return f"{mu:.4f} ± 0.0000"
    return f"{mu:.4f} ± {sigma:.4f}"

# --- STEP 3: GENERATE ABLATION TABLE ---
print("=" * 80)
print(" FEATURE BUDGET ABLATION: P/m vs Full P (For m > 1)")
print("=" * 80)

pivot_ablation = df_ablation.pivot_table(
    index=["dataset", "Subspace_Strategy"], 
    columns=["m", "Feature_Budget"], 
    values="acc", 
    aggfunc=format_mean_std
)
pd.set_option("display.max_rows", None)
display(pivot_ablation)

# --- STEP 4: VISUALIZE THE BUDGET GAP ---
datasets = sorted(df_ablation["dataset"].unique())

for dataset in datasets:
    ds_data = df_ablation[df_ablation["dataset"] == dataset]
    
    # We use catplot to separate by Subspace Strategy cleanly into two panels
    g = sns.catplot(
        data=ds_data, 
        kind="bar",
        x="m", 
        y="acc", 
        hue="Feature_Budget", 
        col="Subspace_Strategy",
        height=4, 
        aspect=1.2,
        palette=["#e74c3c", "#2ecc71"], # Red for Old (starved), Green for New (healthy budget)
        errorbar="sd"
    )
    
    g.fig.suptitle(f"Budget Ablation: {dataset}", y=1.05, fontweight="bold")
    g.set_axis_labels("Ensemble Width (m)", "Accuracy")
    g.set_titles("{col_name}")
    
    # Move legend outside to prevent overlap with bars
    sns.move_legend(g, "center left", bbox_to_anchor=(1, 0.5), title="Budget Type")
    
    plt.show()
```
